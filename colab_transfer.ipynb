```python
#@title 1. 传输参数配置
import os

SRC_PATH = "/content/drive/MyDrive/gakumas_assets" #@param {type:"string"}
REMOTE_DIR = "/colab_transfer" #@param {type:"string"}
PACK_MODE = "single" #@param ["single", "subdir", "direct"]
ARCHIVE_FORMAT = "zip" #@param ["zip", "tar"]
WORKERS = 16 #@param {type:"slider", min:1, max:32, step:1}
CHUNK_MB = 16 #@param [4, 8, 16, 32] {type:"raw"}
CONFLICT = "overwrite" #@param ["overwrite", "rename", "ask"]
SCRIPT_PATH = "/content/drive/MyDrive/Colab Notebooks/gdrive_to_jbox.py" #@param {type:"string"}

os.environ["JBOX_SRC_PATH"] = SRC_PATH.strip()
os.environ["JBOX_REMOTE_DIR"] = REMOTE_DIR.strip()
os.environ["JBOX_PACK_MODE"] = PACK_MODE.strip()
os.environ["JBOX_ARCHIVE_FORMAT"] = ARCHIVE_FORMAT.strip()
os.environ["JBOX_WORKERS"] = str(int(WORKERS))
os.environ["JBOX_CHUNK_MB"] = str(int(CHUNK_MB))
os.environ["JBOX_CONFLICT"] = CONFLICT.strip()
os.environ["JBOX_SCRIPT_PATH"] = SCRIPT_PATH.strip()

assert os.path.exists(os.environ["JBOX_SRC_PATH"]), f"❌ 找不到源路径: {os.environ['JBOX_SRC_PATH']}"
assert os.path.isfile(os.environ["JBOX_SCRIPT_PATH"]), f"❌ 找不到脚本: {os.environ['JBOX_SCRIPT_PATH']}"

print("✅ 传输参数已导入：")
print(f"  SRC_PATH       = {os.environ['JBOX_SRC_PATH']}")
print(f"  REMOTE_DIR     = {os.environ['JBOX_REMOTE_DIR']}")
print(f"  PACK_MODE      = {os.environ['JBOX_PACK_MODE']}")
print(f"  ARCHIVE_FORMAT = {os.environ['JBOX_ARCHIVE_FORMAT']}")
print(f"  WORKERS        = {os.environ['JBOX_WORKERS']}")
print(f"  CHUNK_MB       = {os.environ['JBOX_CHUNK_MB']}")
print(f"  CONFLICT       = {os.environ['JBOX_CONFLICT']}")
print(f"  SCRIPT_PATH    = {os.environ['JBOX_SCRIPT_PATH']}")
```

```python
#@title 2. 一键打包并断点续传至交大云盘 (pan.sjtu.edu.cn)
import importlib.util
import os

AUTH_MODE = "UserToken" #@param ["UserToken", "JAAuthCookie"]
RUN_SPEEDTEST = False #@param {type:"boolean"}
FORCE_REPACK = False #@param {type:"boolean"}
FORCE_UPLOAD = False #@param {type:"boolean"}
KEEP_TEMP = False #@param {type:"boolean"}
DRY_RUN = False #@param {type:"boolean"}

script_path = os.environ.get("JBOX_SCRIPT_PATH", "/content/drive/MyDrive/Colab Notebooks/gdrive_to_jbox.py")
src_path = os.environ.get("JBOX_SRC_PATH", "")
assert src_path, "❌ 请先运行上面的「1. 传输参数配置」代码块！"
assert os.path.isfile(script_path), f"❌ 找不到脚本: {script_path}"

spec = importlib.util.spec_from_file_location("gdrive_to_jbox", script_path)
mod = importlib.util.module_from_spec(spec)
spec.loader.exec_module(mod)

mod.transfer(
    src_path=src_path,
    remote_dir=os.environ.get("JBOX_REMOTE_DIR", "/colab_transfer"),
    pack_mode=os.environ.get("JBOX_PACK_MODE", "single"),
    archive_format=os.environ.get("JBOX_ARCHIVE_FORMAT", "zip"),
    auth_mode=AUTH_MODE,
    workers=int(os.environ.get("JBOX_WORKERS", "16")),
    chunk_mb=int(os.environ.get("JBOX_CHUNK_MB", "16")),
    conflict=os.environ.get("JBOX_CONFLICT", "overwrite"),
    force_repack=FORCE_REPACK,
    force_upload=FORCE_UPLOAD,
    keep_temp=KEEP_TEMP,
    speedtest=RUN_SPEEDTEST,
    dry_run=DRY_RUN,
)
```
